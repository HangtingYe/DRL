in the main results, it is identical to that we only query one time. the mean and std is over random seed

```python
result = result_report_sorted

evaluation = 'AUC-PR'

from predefines import epochs, processors, com_ratios

result = result_report_sorted

seed = 42

prototype_num = 5


for model_type in ['PN']:
    result_ = []
    datanames_dict = dict()
    for dataname in sorted_names:    
        preprocess = processors[dataname]
        
        input_info_ratio, cl_ratio = com_ratios[dataname].split(',')     
        
        try:
            result_dict = np.load(
                f'./results/{dataname}_{model_type}_{preprocess}_True_False_True_{input_info_ratio}_True_{cl_ratio}_{prototype_num}_{seed}.npy',
                allow_pickle=True
            ).item()
            value = result_dict.get(evaluation, np.nan)  # 如果没这个key就填 NaN
        except Exception as e:
            # 文件不存在或读取失败时填 NaN
            value = np.nan
        
        result_.append(value)
    
    result[model_type] = result_
```

```python
mean_row = result.mean()
ranks = result.rank(axis=1, method='average', ascending=False)
rank_row = ranks.mean()
row_max = result.max(axis=1)
is_max = result.eq(row_max, axis=0)
win_count = is_max.sum().astype(int)

# 将均值行追加到数据框末尾，并指定行名为 'Mean'
result.loc['Average'] = mean_row
result.loc['Rank'] = rank_row
result.loc['Win'] = win_count

result
```
